##### מתודות הרמה

```python
x = torch.tensor([1, 2, 3, 4])
y = torch.tensor([5, 6, 7, 8])
print(x+y, x-y, x*y, x/y, sep='\n')
```

```python
x = torch.ones(size=(10**5,))
y = torch.randint(low=0, high=10, size=(10**5,)) # טנזור מ0 ל9 בגודל 100,000
```

```python
%%timeit # מודד כמה זמן לוקח להריץ את כל התא, וחוזר על ההרצה כמה פעמים כדי להעריך את הזמן
x*y # הרצה ע"י פייטורץ מבצע במקביל, מה שמאפשר ניצול של מעבד גרפי במידה וקיים
```

```python
%%timeit
for i in range(x.numel()): # מחזיר את מספר האיברים בטנסור ולכן רץ 100,000 פעמים
    x[i]*y[i] # הרצה ע"י לולאה בפייתון, מה שמאט את הביצועים
```

##### מתודות הפחתה - מתודות סיכום המחזירות סקלר

```python
x = torch.tensor([1, 2, 3, 4])
y = torch.tensor([5, 6, 7, 8])
print(x@y, x==1, sep='\n') # פעולת מכפלה מטריצית (מכפלה פנימית), ובדיקה אם האיבר שווה ל1

# כאשר הווקטור חד מימדי - הסימן @ מחשב את המכפלה הפנימית שלהם
# אם נשנה אותם לווקטור עמודה ווקטור שורה דו מימדיים  - נוכל לקבל את תוצאת כפל המטריצות שלהם
```

```python
x = x.reshape(1, 4) # שינוי צורת הטנסור לווקטור שורה
y = y.reshape(4, 1) # שינוי צורת הטנסור לווקטור עמודה
print(x@y, y@x, sep='\n') # פעולת כפל מטריציות
```

```python
x = torch.ones(size=(3,))
y = torch.ones(size=(3, 3))
z = torch.ones(size=(3, 3, 3))
print(x.norm(), y.min(), z.sum(),sep='\n') # חישוב נורמה, מינימום וסכום של הטנסורים

# תזכורת -נורמה של וקטור זה סכום של הערכים המוחלטים של כל האיברים בווקטור
# במילים אחרות - המרחק של הווקטור מראשית הצירים - זה מדד לאורך שלו
# מעלים כל איבר בריבוע, מחברים את כל הערכים, ולבסוף לוקחים שורש ריבועי של הסכום
```

פעולת הפחתה תמיד תחזיר סקלר, לא משנה כמה מימדית יש בקלט
עם זאת ניתן להעביר רק את אחד המימדים של הטנסור, ואז נקבל טנסור עם מימד אחד פחות מהקלט

```python
x = torch.arange(24).reshape(2, 3, 4)
print(x, x.size(), sep='\n')
print()

y = x.sum()
print(y, y.size(), sep='\n')
print()

z = x.sum(dim=0) # סכום לאורך המימד הראשון (שורות)
print(z, z.size(), sep='\n')
print()

w = x.sum(dim=1) # סכום לאורך המימד השני (עמודות) - בכל מטריצה בוצע סכום לעמודות
print(w, w.size(), sep='\n')
```

torch.rand(...) — מספרים עשרוניים מהתפלגות אחידה בטווח [0, 1).

torch.randn(...) — מספרים עשרוניים מהתפלגות נורמלית סטנדרטית: ממוצע 0 וסטיית תקן 1.

torch.randint(low, high, ...) — מספרים שלמים אקראיים בטווח [low, high)

```python
a = torch.rand(size=(100,2,2))

# א. ערכים עצמיים של כל המטריצות
eigenvalues = torch.linalg.eigvals(a) # חישוב ערכי עצמי של מטריצה אקראית בגודל 2 על 2, עבור כל מטריצה ב100 מטריצות
# התוצאה היא בגודל (100, 2) — שני ערכים עצמיים לכל מטריצה. המשתנה עשוי להחזיר ערכים מרוכבים, אך כאן כל איברי המטריצות בין 0 ל־1
```

```python
# ב. נורמת פרובניוס של כל מטריצה
# נורמת פרובניוס היא השורש של סכום ריבועי כל האיברים במטריצה
frobenius_norms = torch.linalg.matrix_norm(a, ord="fro", dim=(-2, -1))
print(frobenius_norms, frobenius_norms.size(), sep='\n')
# מתקבל וקטור בגודל (100), עם נורמה אחת לכל מטריצה
```

```python
# ג. הנורמה האוקלידית של הטנזור כולו
tensor_norm = a.norm()
print(tensor_norm, tensor_norm.size(), sep='\n')
```

```python
# ד. חשבו את נורמת אינסוף של שתי המטריצות בגודל 100,2 המתקבלות עבור כל אחד מערכי המימד האחרון
A = a[:, :, 0] # נבחר את כל המטריצות בגודל 2 על 2, ונבחר את הערך הראשון של כל מטריצה (מימד אחרון)
B = a[:, :, 1] # נבחר את כל המטריצות בגודל 2 על 2, ונבחר את הערך השני של כל מטריצה (מימד אחרון)

norm_A = torch.linalg.matrix_norm(A, ord=float("inf"))
norm_B = torch.linalg.matrix_norm(B, ord=float("inf"))

# נורמת האינסוף של מטריצה היא סכום הערכים המוחלטים הגדול ביותר של שורה.
# לכן פייטורץ' מחבר את הערכים של כל שורה ומחזיר את הערך הגדול ביותר מבין כל השורות
# התוצאה של כל נורמה זה מספר יחיד, ולכן מתקבל משתנה בגודל () ולא וקטור

print(norm_A, norm_B, sep='\n')
```
