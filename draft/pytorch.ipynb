##### מתודות הפחתה - מתודות סיכום המחזירות סקלר

```python
x = torch.tensor([1, 2, 3, 4])
y = torch.tensor([5, 6, 7, 8])
print(x@y, x==1, sep='\n') # פעולת מכפלה מטריצית (מכפלה פנימית), ובדיקה אם האיבר שווה ל1

# כאשר הווקטור חד מימדי - הסימן @ מחשב את המכפלה הפנימית שלהם
# אם נשנה אותם לווקטור עמודה ווקטור שורה דו מימדיים  - נוכל לקבל את תוצאת כפל המטריצות שלהם
```

```python
x = x.reshape(1, 4) # שינוי צורת הטנסור לווקטור שורה
y = y.reshape(4, 1) # שינוי צורת הטנסור לווקטור עמודה
print(x@y, y@x, sep='\n') # פעולת כפל מטריציות
```

```python
a = torch.rand(size=(100,2,2))

# א. ערכים עצמיים של כל המטריצות
eigenvalues = torch.linalg.eigvals(a) # חישוב ערך עצמי של מטריצה 2*2 אקראית, עבור כל מטריצה ב100 מטריצות
# התוצאה היא בגודל (100, 2) — שני ערכים עצמיים לכל מטריצה. המשתנה עשוי להחזיר ערכים מרוכבים, אך כאן כל איברי המטריצות בין 0 ל־1
```

```python
# ב. נורמת פרובניוס של כל מטריצה
# נורמת פרובניוס היא השורש של סכום ריבועי כל האיברים במטריצה
frobenius_norms = torch.linalg.matrix_norm(a, ord="fro", dim=(-2, -1))
print(frobenius_norms, frobenius_norms.size(), sep='\n')
# מתקבל וקטור בגודל (100), עם נורמה אחת לכל מטריצה
```

```python
# ג. הנורמה האוקלידית של הטנזור כולו
tensor_norm = a.norm()
print(tensor_norm, tensor_norm.size(), sep='\n')
```

```python
# ד. חשבו את נורמת אינסוף של שתי המטריצות בגודל 100,2 המתקבלות עבור כל אחד מערכי המימד האחרון
A = a[:, :, 0] # נבחר את כל המטריצות בגודל 2 על 2, ונבחר את הערך הראשון של כל מטריצה (מימד אחרון)
B = a[:, :, 1] # נבחר את כל המטריצות בגודל 2 על 2, ונבחר את הערך השני של כל מטריצה (מימד אחרון)

norm_A = torch.linalg.matrix_norm(A, ord=float("inf"))
norm_B = torch.linalg.matrix_norm(B, ord=float("inf"))

# נורמת האינסוף של מטריצה היא סכום הערכים המוחלטים הגדול ביותר של שורה.
# לכן פייטורץ' מחבר את הערכים של כל שורה ומחזיר את הערך הגדול ביותר מבין כל השורות
# התוצאה של כל נורמה זה מספר יחיד, ולכן מתקבל משתנה בגודל () ולא וקטור

print(norm_A, norm_B, sep='\n')
```

##### גזירה אוטומטית

```python
x = torch.tensor([2.], requires_grad=True)

# כאשר עושים על הטנזור שהוא דורש גזירה - אנו מצהירים על הטנזור כמשתנה שבהמשך נרצה לגזור
# לאחר מכן פעולה הגזירה מתבצעת על ידי פקודת backward() 
# הנגזרת תישמר במאפיין בשם גראד
```

```python
y = x**2
z = y + x # =>  z=x^2+x =>   z'=2x+1 
z.backward()
print(x.grad) # יוחזר 5 לאחר הצבת איקס שווה 2
```

```python
# נגזרות חלקיות
x = torch.tensor([2.,3.], requires_grad=True)
y = x[0]*x[1]
y.backward()
print(x.grad)
```

<div dir="rtl">
הסיבה שהפלט הוא 3,2 זה כי כדי לגזור את הפונקציה y גוזרים בשלב ראשון לפי x[0], כלומר מתייחסים אליו כאל המשתנה x, ולx[1] בתור המקדם של x. לכן גזירה שלו תחזיר 3.
באותו אופן גזירה של y לפי x[1] - נתייחס אליו בתור המשתנה, וx[0] בתור המקדם, והוא שווה 2, ולכן הגזירה תחזיר 2
</div>

```python
x = torch.tensor([2.,3.],requires_grad=True)
y = torch.tensor([4.,5.],requires_grad=True)
w = x.sum()**2
z = w + y.sum()**3+x[0]+y[1]
z.backward()
print(x.grad, y.grad, sep='\n')
```

<div dir="rtl">
בדוגמה זו אנו עוקבים אחרי פעולות החשבון שמבוצעות על שני טנזורים שונים בו־זמנית, ובעת הרצת המתודה <code dir="ltr">backward()</code> מחושב בו־זמנית הגרדיאנט של <code dir="ltr">z</code> ביחס לכל אחד מהם.
<br><br>
<b>חישוב הגרדיאנטים בקצרה:</b>
<ul>
  <li><b>עבור <code dir="ltr">x</code> (כאשר סכום איבריו הוא <code dir="ltr">5</code>):</b> נגזרת <code dir="ltr">w</code> לפי כל איבר היא <code dir="ltr">2*(2+3)=10</code>. מכיוון ש-<code dir="ltr">x[0]</code> מופיע גם כאיבר בודד ב-<code dir="ltr">z</code>, מתווסף לו <code dir="ltr">1</code> ומתקבל <code dir="ltr">[11., 10.]</code>.</li>
  <li><b>עבור <code dir="ltr">y</code> (כאשר סכום איבריו הוא <code dir="ltr">9</code>):</b> נגזרת החזקה השלישית לפי כל איבר היא <code dir="ltr">3*(4+5)**2=243</code>. מכיוון ש-<code dir="ltr">y[1]</code> מופיע גם כאיבר בודד ב-<code dir="ltr">z</code>, מתווסף לו <code dir="ltr">1</code> ומתקבל <code dir="ltr">[243., 244.]</code>.</li>
<li>נדגיש כי אם מבצעים חישובי נגזרת בלולאה - אם רוצים לחשב כל נגזרת בנפרד ולא לבצע סכימה של הכל - צריך לאפס את הגרדיאנט על ידי x.grad = None</li>
</ul>
</div>


```python
x = torch.tensor(2., requires_grad=True)
y = x**2

y.backward(create_graph=True)

t = x.grad
x.grad = None
t.backward()

print(x.grad)  # מדפיס את הנגזרת השנייה (tensor(2.))
```

<div dir="rtl">
<ul><li>בלי create_graph (ברירת מחדל): כאשר מחשבים y.backward(), הערך שנשמר ב-x.grad הוא טנזור רגיל ללא היסטוריית חישובים (grad_fn=None). אם תנסה לגזור את x.grad שוב, PyTorch לא יידע איך x.grad תלוי ב-x</li>
<li>עם create_graph=True: הפעולה y.backward(create_graph=True) הופכת בעצמה לחלק מגרף החישוב. הטנזור שמתקבל ב-x.grad שומר בתוכו פונקציית גזירה (grad_fn), כך שניתן להפעיל עליו .backward() פעם נוספת.
</li></ul></div>
